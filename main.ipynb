```python
import os
from groq import Groq
from langchain_community.tools.tavily_search import TavilySearchResults
from IPython.display import Markdown, display
from dotenv import load_dotenv
import warnings
warnings.filterwarnings('ignore')

load_dotenv()

# Clients
groq_client = Groq(api_key=os.getenv("GROQ_API_KEY"))
tavily_search = TavilySearchResults(max_results=5)

# Search + Answer function
def ask_agent(question: str) -> str:
    # Step 1: Search the web
    search_results = tavily_search.run(question)
    context = str(search_results)[:8000]

    # Step 2: Ask Groq to summarize in Markdown
    response = groq_client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": "You are a helpful research assistant. Use the provided search results to answer the question in structured Markdown format."},
            {"role": "user", "content": f"Search results:\n{context}\n\nQuestion: {question}"}
        ],
        temperature=0.5,
        max_tokens=1024
    )
    return response.choices[0].message.content


```

### Exercise 1: AI in Healthcare

Goal: Investigate and summarize the latest advancements in generative AI applied to healthcare in 2025.

- Design a prompt that asks the agent to retrieve the most recent updates.
- Ensure the agent outputs a structured response in Markdown.

```python
# Your prompt here
prompt_1 = """
Search the web for the latest advancements in generative AI applied to healthcare in 2025.
Return a structured Markdown response with the following sections:
- ## Key Advancements
- ## Notable Companies or Research Groups
- ## Impact on Patients and Clinicians
- ## Sources
"""

# Run it
response_1 = ask_agent(prompt_1)
```

```python
display(Markdown(response_1))
```

### Exercise 2: AI Startups Landscape

Goal: Track 2025’s top emerging AI startups and their innovations.

- Create a prompt that instructs the agent to deliver a clean Markdown summary.
- Tip: Ask for company names, product highlights, and sources.

```python
# Your prompt here
prompt_2 = """
Search the web for the top emerging AI startups of 2025.
Return a structured Markdown response with the following sections:
- ## Top AI Startups of 2025
For each startup include: company name, main product or innovation, and source URL.
- ## Trends Across These Startups
- ## Sources
"""

# Run it
response_2 = ask_agent(prompt_2)
```

```python
display(Markdown(response_2))
```
